# Truth Table Generator
This notebook generates truth tables for logical expressions in propositional logic.


```python
import ipywidgets as widgets
from IPython.display import display

# Create a text box for the logical expression
expr_input = widgets.Text(
    value='A | ~A',  # default expression
    description='Expression:',
    layout=widgets.Layout(width='400px')
)

display(expr_input)
```

```python
import re

# Function to find variables in the expression
def get_variables(expr):
    return sorted(list(set(re.findall(r'\b[A-Z]\b', expr))))

# Example usage
variables = get_variables(expr_input.value)
print("Variables in expression:", variables)
```

```python
import itertools
import pandas as pd

# Function to generate the truth table
def generate_truth_table(expr):
    vars = get_variables(expr)
    table = []
    for values in itertools.product([0,1], repeat=len(vars)):
        env = {var: bool(val) for var,val in zip(vars, values)}
        result = eval(expr, {}, env)  # evaluate expression
        table.append(list(values) + [int(result)])
    columns = vars + ['Result']
    return pd.DataFrame(table, columns=columns)

# Example
df = generate_truth_table(expr_input.value)
df
```

```python
# Function to check if expression is Tautology, Contradiction, or Contingent
def check_expression_type(df):
    if df['Result'].all():
        return "Tautology"
    elif not df['Result'].any():
        return "Contradiction"
    else:
        return "Contingent"

check_expression_type(df)
```

```python
# Button to generate truth table
button = widgets.Button(description="Generate Truth Table")

def on_click(b):
    df = generate_truth_table(expr_input.value)
    display(df)
    print("Expression Type:", check_expression_type(df))

button.on_click(on_click)
display(button)
```
